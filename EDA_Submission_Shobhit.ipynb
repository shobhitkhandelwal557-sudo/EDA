## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# Removing duplicates across the dataset
data1_cleaned = data1.drop_duplicates()
data1_cleaned
````

````python
# 1. Calculate Q1, Q3, and IQR for all numeric columns automatically
numeric_data = data1_cleaned.select_dtypes(include=['number'])

# 2. Run the outlier check only on those numeric columns

Q1 = numeric_data.quantile(0.25, numeric_only=True)
Q3 = numeric_data.quantile(0.75, numeric_only=True)
IQR = Q3 - Q1


# 2. Create a rule map (True if a cell is an outlier)
is_outlier = (numeric_data < (Q1 - 1.5 * IQR)) | (numeric_data > (Q3 + 1.5 * IQR))
outlier_counts = is_outlier.sum()
print(outlier_counts)
````

````python
# Representing outliers using box plot

# 1. Select only the columns that actually have outliers (count > 0)
columns_with_outliers = outlier_counts[outlier_counts > 0].index.tolist()

# 2. Set up a grid layout dynamically based on the number of columns
num_cols = len(columns_with_outliers)
fig, axes = plt.subplots(nrows=(num_cols + 2) // 3, ncols=3, figsize=(16, num_cols * 1.5))
axes = axes.flatten()

# 3. Loop through and plot each column
for i, col in enumerate(columns_with_outliers):
    sns.boxplot(x=data1_cleaned[col], ax=axes[i], color='skyblue')
    axes[i].set_title(f'{col} (Outliers: {outlier_counts[col]})')
    axes[i].set_xlabel('')

# 4. Hide any unused empty subplot slots in the grid
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
````

````python
#  23000 are outliers. Cap outliers instead of removing rows
data1_final = data1_cleaned.copy()

for col in columns_with_outliers:
    lower_bound = Q1[col] - 1.5 * IQR[col]
    upper_bound = Q3[col] + 1.5 * IQR[col]
    data1_final[col] = data1_final[col].clip(lower=lower_bound, upper=upper_bound)

print(f"Original dataset rows: {data1_cleaned.shape[0]}")
print(f"Final dataset rows: {data1_final.shape[0]}")  # will still be 100000
````

````python
# Box plot to confirm outliers have been handled successfully in the dataset

# 1. Recheck which columns still have outliers after capping
Q1_final = data1_final.select_dtypes(include=['number']).quantile(0.25, numeric_only=True)
Q3_final = data1_final.select_dtypes(include=['number']).quantile(0.75, numeric_only=True)
IQR_final = Q3_final - Q1_final

numeric_data_final = data1_final.select_dtypes(include=['number'])
is_outlier_final = (numeric_data_final < (Q1_final - 1.5 * IQR_final)) | (numeric_data_final > (Q3_final + 1.5 * IQR_final))
outlier_counts_final = is_outlier_final.sum()

print(outlier_counts_final)

# 2. Plot box plots for the SAME columns as before, so you can compare before/after
num_cols = len(columns_with_outliers)
fig, axes = plt.subplots(nrows=(num_cols + 2) // 3, ncols=3, figsize=(16, num_cols * 1.5))
axes = axes.flatten()

for i, col in enumerate(columns_with_outliers):
    sns.boxplot(x=data1_final[col], ax=axes[i], color='lightgreen')
    axes[i].set_title(f'{col} (Outliers: {outlier_counts_final[col]})')
    axes[i].set_xlabel('')

for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
````

````python
import pandas as pd
import matplotlib.pyplot as plt

df = data1_final
# ---- Config ----
rate_col = "Interest_Rate"
enq_col = "Num_Credit_Inquiries"
n_bins = 6
# ----------------

# Clean column names (removes stray spaces)
df.columns = df.columns.str.strip()

# Make both columns numeric
for c in [rate_col, enq_col]:
    df[c] = pd.to_numeric(df[c], errors="coerce")
data = df[[rate_col, enq_col]].dropna().copy()

# Bin the interest rate and average enquiries per bin in one step
bins = pd.cut(data[rate_col], bins=n_bins)
avg = data.groupby(bins, observed=True)[enq_col].mean().round(2)

labels = [f"{int(iv.left)}-{int(iv.right)}" for iv in avg.index]

fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.bar(labels, avg.values, color="#3498db", edgecolor="white")
ax.bar_label(bars, fmt="%.2f", padding=3)

ax.set_title(f"Average {enq_col} by {rate_col} Range")
ax.set_xlabel(f"{rate_col} (range)")
ax.set_ylabel(f"Average {enq_col}")
ax.grid(axis="y", linestyle="--", alpha=0.4)
fig.tight_layout()
plt.show()
````

### Chart-17

````python
#Plot to show the distribution of average number of Delayed payments over Average number of Bank accountd

df = data1_final   # or use your existing DataFrame

# ---- Config ----
x_col = "Num_Bank_Accounts"
y_col = "Num_of_Delayed_Payment"
min_customers = 100          # hide x values with too few customers
max_x = 12                   # drop extreme or invalid account counts
# ----------------

df.columns = df.columns.str.strip()
for c in [x_col, y_col]:
    df[c] = pd.to_numeric(
        df[c].astype(str).str.replace(r"[^0-9.\-]", "", regex=True),
        errors="coerce",
    )
data = df[[x_col, y_col]].dropna()
data = data[(data[x_col] >= 0) & (data[x_col] <= max_x)]

# Average delayed payments for each number of bank accounts
g = data.groupby(x_col)[y_col].agg(["mean", "count", "sem"])
g = g[g["count"] >= min_customers]
ci = 1.96 * g["sem"]                      # 95% confidence band

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(g.index, g["mean"], marker="o", linewidth=2.5, markersize=8,
        color="#e74c3c", label="Average delayed payments")
ax.fill_between(g.index, g["mean"] - ci, g["mean"] + ci,
                color="#e74c3c", alpha=0.15, label="95% confidence band")

# Optional straight trend line
slope, intercept = np.polyfit(g.index, g["mean"], 1)
ax.plot(g.index, slope * g.index + intercept, linestyle="--",
        color="gray", label="Trend line")

# Data labels
for x, y in zip(g.index, g["mean"]):
    ax.annotate(f"{y:.1f}", (x, y), textcoords="offset points",
                xytext=(0, 10), ha="center", fontsize=9)

ax.set_xticks(g.index)
ax.set_title("Average Delayed Payments by Number of Bank Accounts", fontsize=14)
ax.set_xlabel("Number of bank accounts")
ax.set_ylabel("Average Num_of_Delayed_Payment")
ax.grid(True, linestyle="--", alpha=0.4)
ax.set_axisbelow(True)
ax.legend()
fig.tight_layout()
plt.show()

# Exact values and customers per point
print(g.round(2))
print("Correlation:", round(data[x_col].corr(data[y_col]), 2))
````

### Chart 19- Line Chart

````python
df = data1_final

# ---- Config ----
income_col = "Annual_Income"
spend_col = "Amount_spent_monthly"   # created below if it doesn't exist
n_bands = 10
# ----------------

df.columns = df.columns.str.strip()

def to_num(s):
    return pd.to_numeric(s.astype(str).str.replace(r"[^0-9.\-]", "", regex=True), errors="coerce")

# Derive spend if the column isn't in the data
if spend_col not in df.columns:
    for c in ["Monthly_Inhand_Salary", "Amount_invested_monthly", "Total_EMI_per_month", "Monthly_Balance"]:
        df[c] = to_num(df[c])
    df[spend_col] = (df["Monthly_Inhand_Salary"] - df["Amount_invested_monthly"]
                     - df["Total_EMI_per_month"] - df["Monthly_Balance"])

df[income_col] = to_num(df[income_col])
df[spend_col] = to_num(df[spend_col])
data = df[[income_col, spend_col]].dropna()
data = data[data[spend_col] >= 0]          # drop impossible negative spend

# Equal-sized income bands, average of BOTH columns in each band
data["band"] = pd.qcut(data[income_col], q=n_bands, duplicates="drop")
g = data.groupby("band", observed=True).agg(
    avg_income=(income_col, "mean"),
    avg_spend=(spend_col, "mean"),
    customers=(income_col, "size"),
).round(2)

fig, ax = plt.subplots(figsize=(9, 5.5))
ax.plot(g["avg_income"], g["avg_spend"], marker="o", markersize=8,
        linewidth=2.5, color="#8e44ad")
for x, y in zip(g["avg_income"], g["avg_spend"]):
    ax.annotate(f"{y:,.0f}", (x, y), textcoords="offset points",
                xytext=(0, 10), ha="center", fontsize=9)

slope, intercept = np.polyfit(g["avg_income"], g["avg_spend"], 1)
xs = np.linspace(g["avg_income"].min(), g["avg_income"].max(), 50)
ax.plot(xs, slope * xs + intercept, "--", color="gray", label="Trend line")

ax.set_title("Average Monthly Spend vs Average Annual Income (by income band)", fontsize=13)
ax.set_xlabel("Average Annual_Income (per band)")
ax.set_ylabel("Average estimated monthly spend")
ax.grid(True, linestyle="--", alpha=0.4)
ax.legend()
fig.tight_layout()
plt.show()

print(g)
print("Correlation of band averages:", round(g["avg_income"].corr(g["avg_spend"]), 2))
````

I picked the specific line chart to analyze the variation of Average estimated monthly spend with the increase of average annual income. As, these two columns are continuous data so line chart is easy to observe.

####What is/are insights found from the chart?

*   There are total 10000 customers in the highest income band with average annual income around Rs. 131539.37 and they spend Rs. 10050.36 per month that is around 91.7%.

### Chart-20

````python


df = data1_final   # assigning final dataset to df

# ---- Config ----
group_col = "Credit_Score"
loan_col = "Num_of_Loan"
hist_col = "Credit_History_Age"      # in months
order = ["Poor", "Standard", "Good"]
colors = {"Poor": "#e74c3c", "Standard": "#f39c12", "Good": "#27ae60"}
# ----------------

df.columns = df.columns.str.strip()
for c in [loan_col, hist_col]:
    df[c] = pd.to_numeric(
        df[c].astype(str).str.replace(r"[^0-9.\-]", "", regex=True),
        errors="coerce",
    )

avg = df.groupby(group_col)[[loan_col, hist_col]].mean().reindex(order).round(2)

fig, ax1 = plt.subplots(figsize=(9, 5.5))

# Columns: average loans (left axis)
bars = ax1.bar(avg.index, avg[loan_col], color=[colors[g] for g in avg.index],
               width=0.45, label="Avg number of loans")
ax1.bar_label(bars, fmt="%.2f", padding=3)
ax1.set_ylabel("Average Num_of_Loan")
ax1.set_ylim(0, avg[loan_col].max() * 1.25)
ax1.set_xlabel("Credit_Score")

# Line: average credit history age (right axis)
ax2 = ax1.twinx()
ax2.plot(avg.index, avg[hist_col], color="black", marker="o", linewidth=2.5,
         markersize=8, label="Avg credit history age (months)")
for x, y in zip(avg.index, avg[hist_col]):
    ax2.annotate(f"{y:.1f}", (x, y), textcoords="offset points",
                 xytext=(0, 10), ha="center", fontsize=10)
ax2.set_ylabel("Average Credit_History_Age (months)")
ax2.set_ylim(0, avg[hist_col].max() * 1.25)

h1, l1 = ax1.get_legend_handles_labels()
h2, l2 = ax2.get_legend_handles_labels()
ax1.legend(h1 + h2, l1 + l2, loc="upper center", bbox_to_anchor=(0.5, -0.12), ncol=2, frameon=False)

ax1.set_title("Average Number of Loans and Credit History Age by Credit Score", fontsize=13)
ax1.grid(axis="y", linestyle="--", alpha=0.4)
ax1.set_axisbelow(True)
fig.tight_layout()
plt.show()

print(avg)
````
